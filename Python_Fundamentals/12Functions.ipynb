A function in Python is a 
- reusable
- named block of code designed to perform a single, well defined task

Instead of duplicating the same logic acoss multiple files or API endpoints, we can define a function once with def and call it whenever needed.

## Defining and Calling a function

A function definition consissts of the def keyword a snake_case name, paranthesis (), a colon: and an indented body.

Defining a function only registers it in memory - the code inside does not run unti we call it using function_name()

```python
# Defining a basic function

def print_health_status() -> None:
    print("Sevice status: 200 OK")

# calling the function
print_health_status()
print_health_status()
```

 ## Type annotations and Docstrings

 in Modern Pythong and FatAPI codebases, functions include type hints on inputs and outputs (eg: -> str or -> None) alongside a triple quoted docstring ("""..""") right below the def line to explain what the function does

```python
def format_log_entry(service: str, status_code: int) -> str:
    """ Formats an HTTP event into a stardized log line"""

    return f"[{service.upper()}] HTTP {status_code}"


entry = format_log_entry("auth-api",201)
print(entry)
```

# Functions are first class objects

In Python, functions are objects just like strings or integers. 

We can assign a function to another variable, store functions inside a dictionary or list, and pass a function into another function without calling it - (by omitting the paranthesis)

```python
def handle_get() -> str:
    return "Fetching resources..."

def handle_post() -> str:
    return "Creating resource..."

# Mapping HTTP method strings directly to handler functions

route_handlers = {
    "GET" :handle_get,
    "POST": handle_post,
}

# Look up the function by key, then call it with ()

incoming_method = "POST"
handler = route_handlers[incoming_method]
print(handler())
```

## Real world backend and API examples

in FastAPI , every single API endpoint is literally a python function registered to a URL path, supported by small helper functions for validations and formatting

```python
# Example A - Input sanitization helper

def normalize_email(raw_email : str) -> str:
    """ Strips surrounding whitespace and converts to lowercase"""

    return raw_email.strip().lower()

clean_email = normalize_email("  Devanath.DR@Company.COM   ")
print("Normalized:", clean_email)
```

```python
# Example B  - Simulated FastAPI endpoint handler

def get_system_info() -> dict[str,str | int] :
    """ SImulates a GET /health endpoint returning a JSON-ready dictionary"""

    return {
        "service" : "user-management-api",
        "version" : "1.0.0",
        "status" : "healthy",
        "uptime_seconds" : 3600,
    }

response_payload = get_system_info()
print("API Response:",response_payload)
```
