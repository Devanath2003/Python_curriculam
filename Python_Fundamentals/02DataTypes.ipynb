Python figures out data types automatically when we assign a value.

## Five Core Primitive Data Types

- int (Integer) : Whole numbers, Positive or negative , without decimals (32,-19,0)
- float(Floating point) : Numbers containing a decimal point(10.2,-0.5, 3.0)
- str (Strin): Sequences of text characters wrapped in single or double quotes
- bool(Boolean) : Logical values representing truth states- strictly True or False
- NoneType(None): Represents the intentional absence of value or an empty state (equivalent to null in databases and JSON)

```python
#Basic data type examples

user_age = 24
account_balance = 1234.55
user_name = "Devanath"
is_verified = True
middle_name = None

# using built it type() function to inspect any variable
print(type(user_age))
print(type(account_balance))
print(type(user_name))
print(type(is_verified))
print(type(middle_name))
```

## Type Annotations (Type Hints)

No need to declare types, so a variable can change types if reassigned.

Modern python and FastAPI paticularly uses **Type Annotations**

variable_name : type = value -> it is for document what type a variable expected to hold.

<u> Python ignores them at runtime, while editors and frameworks use them to catch bugs early.</u>


```python
# Standard dynamic assignment
port = 8000

# Annotated assignment (prefered in FastAPI codebases)
max_connections: int = 100
timeout_seconds: float = 2.5
environment: str = "production"
enable_cache : bool = True
```
