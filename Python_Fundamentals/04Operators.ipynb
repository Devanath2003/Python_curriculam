OPerators are special symbols and keywords in Python that perform mathematical calculations, update variables, compare values, and combine logical conditions

## Arithmetic & assignment operators

arithmatic operators handle math, while ocmpound assingment operators (+=, -=) etc update a variable in place

```python
a = 10 
b = 3

print(a+b)
print(a-b)
print(a*b)
print(a/b) # True division - always returns a float
print(a//b) # Floor division - drops reminder - return int
print(a%b) # Modulo - returns the reminder
print(a**b) # Exponentiation or power

# Compound assingment 
requests_count = 0
requests_count += 1 # req_count = Req_count + 1
requests_count *= 5 # req_count = req_count * 5
print(requests_count)
```

## Comparison and logical operators

It evaluates relationships bw values and always return a bool

Logical operators (and, or , not) combine multiple broken expressions.

```python
# Comparison operators 

status = 200
print(status == 200) # Equal to
print(status != 499)  # NOt equal to
print(status >= 132) # Greater than or euqal to
print(status < 300) # Less than

# Logical operators

is_authenticated = True
is_admin = False
is_suspened = False

# and - requires both sides to be true
print(is_authenticated and is_admin)

# or - requires atleast on side to be true
print(is_admin or is_authenticated)

# not flips 
print(is_authenticated and not is_admin)
```

## Identity operator (is) and Membership (in) operators

- is / is not = checks whether two variable point to the exact same object in memory.

 In profesional Python, always use "is None" or "is not None" instead of == None

- in/ not in = checks whether a value exists inside a sequence (such as a substring inside a str)

```python
api_key = None
print(api_key is None)
print (api_key is not None)

email = "devanath@abc.com"
print("@abc.com" in email)
print("spam" not in email)
```
