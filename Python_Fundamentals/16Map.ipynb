map() is a built in Python function that applies a transformation function to every single item in an iterable (like a list or tuple) and gives back the transformed results

- Think of map() - like a factory conveyor belt: raw items enter on the left, every item passes through the exact same machine (your function), and the converted items come out on the right

### syntax and How works

map(function, iterable)


- function : A lambda or a regular def function (passed without paranthesis)
- iterable : A list, tuple, set, or string whose items we want to transform


**Crucial Rule(Lazy Evaluation)** : map() does not immediately build a list in memory. It returns a special map object (an iterator) that computes items on demand

To see or reuse the results, we need to wrap it inside list()

```python
numbers = [1,2,3,4,5]

# without list(), printing map() only shows its memory address:
result_obj = map(lambda n: n * 10, numbers)
print(result_obj)

# Wrapping in list() executes the converyor belt and collects all items:
multiplied = list(map(lambda n: n*10, numbers))
print(multiplied)
```

## Variety of Basic map() Examples

Example - A : Using BUilt in functions Directly (No lambda needed!)

 Whenever an existing python function already takes one argument, pass its name straight to map()



```python
# converting a list of raw string numbers (eg - from CLI args or csv) into integers
raw_ids = ["101","102","103","104","105"]
numeric_ids = list(map(int,raw_ids))
print(numeric_ids)


# Measuring the character length of every word
routes = ["asb", "asdff","234","dfwhsdfhsdf"]
lengths = sorted(list(map(len,routes)))
print(lengths)
```

Example B : Using a Namde def Function without map()

when the transformatin logic needs a couple of steps, define a clean def function and pass it into map()

```python
def apply_gst(base_price: float) -> float:
    """Adds 17% GST and rounds to 2 decimal places."""
    return round(base_price * 1.18, 2)

cart_prices = [100.0, 250.5, 499.0]
final_prices = list(map(apply_gst, cart_prices))
print(final_prices)
```

Example C : Cleaning and Normalizing Text Data

```python

messy_emails = [" Alice@Example.COM", "BOB#COMPANY.ORG\n", "  charlie@Test.in"]

clean_emails = list(map(lambda email: email.strip().lower(),messy_emails))

print(clean_emails)
```

Example D : Passing Multiple Lists to map() Simultaneously

If our function takes two arguments, we can pass two lists into map()

Python pairs them up element by element and stops as soon as the shorted list ends

```python
base_amounts = [1000,2000,1500]
discounts = [100,250,50]

net_totals = list(map(lambda base, disc: base-disc, base_amounts, discounts))
print(net_totals)
```

## Real worls Backend and API Examples

When a database query returns full user dictionaries, we often need just a list of their email address to pass to a notification service

```python
db_users = [
    {"id": 1, "username": "devanath", "email": "devanath@company.com"},
    {"id": 2, "username": "alice", "email": "alice@company.com"},
    {"id": 3, "username": "bob", "email": "bob@company.com"},
]

recipient_emails = list(map(lambda user: user["email"],db_users)) #the lambda function will extract the email and only returs each of it and map function collects all
print("Send newsletters to:", recipient_emails)
```

## One Important Gotcha : Iterators are single use

Because map() returns an iterator, once you loop over it or convert it to a list() once, it becomes empty - just like water draining out of a tank

Always save list(map(..)) into a variable if we plan to inspect it multiple times

```python
nums = [1,2,3]
doubled_iter = map(lambda x: x*2, nums)
print("First pass: ", list(doubled_iter)) # we will get output for this
print("Second pass", list(doubled_iter)) # this won't cause the iterator output of map already exhausted or become empty
```
