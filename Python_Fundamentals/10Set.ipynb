A set in Python is an 
- Unordered
- mutable collection of 
    - unique
    - immutable items



It is like a mathematical set:
- duplicates are automatically eliminated
- there is no guaranteed index order
- checking whether an items exists inside a set is near-instaneours O(1) lookup speed

##  Creation and Automatic Deduplication

Sets are created using curly braces {}, or the set() constructor.

Important : Empty curly braces {} create an empty dictionary, so we must use set() to create an empty set.

```python
# Duplicates are automcatically stripped out

tags: set[str] = {"python","fastapi","python","postgres","fastapi"}

print(tags)


#Coverting a list with duplicates into a deduplicated set

raw_ids = [101, 102, 102, 105, 101, 108]
unique_ids = set(raw_ids)
print(unique_ids) 

# Creating an empty set 
empty_set: set[str] = set()

```

## Adding adn Removing items

Pay close attention to the difference between .remove and .discard()

using .discard() avoids crashing our API if an item was already deleted

```python
active_sessions = {"sess_1","sess_2","sess_3"}

# Adding items

active_sessions.add("sess_4")
active_sessions.update(["sess_5","sess_1"]) # Adds multiple - ignores duplicates.


print(active_sessions)

# Removing items

active_sessions.remove("sess_2") # Remove items - raise KeyError if missing!
active_sessions.discard("sess_99") # Removes item safely; does NOTHING if missing.

print(active_sessions)
```

## Mathematical Set Operations

sets are better when comparing two collections to find overlaps , differences , or missing permissions.

```python
required_scopes = {"read:users", "write:users","delete:users"}
user_scopes = {"read:users","write:users","read:billing"}

# Intersection (&): Items present in BOTH sets

common = required_scopes & user_scopes
print("Granted scopes area:",common)

# Difference (-): Items in the first set that are MISSING from the second
missing = required_scopes - user_scopes
print("Missng scope",missing)

# Union (|) All unique items combined from both sets

all_scopes = required_scopes | user_scopes
print("Combined scopes:", all_scopes)

# Subset checkz (<=) : Does user have ALL required scopes ?

has_full_access = required_scopes <= user_scopes
print("Full access:", has_full_access)
```

# Real world bakcend and API examples

In FastAPI backends, sets are heavily used for Role-Based Access Control (RBAC), rate-limiting blocklists, and filtering duplicate IDs before hitting PostgreSQL

```python
# Example A: Fast IP Blocklist Lookup

# Checking 'in' on a set is vastly faster than on a large list

BLOCKED_IPS = {"192.168.1.45","10.0.0.99","172.16.0.12"}
client_ip = "10.0.0.99"

if client_ip in BLOCKED_IPS:
    print("403 Forbidden: IP address is blocklisted.")


```

```python
# Example B : Detecting New vs Removed Tags on a Resource Update

existing_post_tags = {"backend", "python", "api"}
updated_post_tags = {"python", "api", "fastapi", "docker"}

tags_to_insert =updated_post_tags - existing_post_tags 
tags_to_delete = existing_post_tags - updated_post_tags

print("Insert into DB:", tags_to_insert)
print("Delete from DB:", tags_to_delete)
```
