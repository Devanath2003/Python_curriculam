## Naming rules and Industry conventions

- Valid characters : only use letters (a-z , A-Z) or numbers 0-9 , dont start with numbers or spaces
- Case sensitivity : user_id, User_id , USER_ID all these are treated differently
- Reserved words: in built keywords like from, import, def , return et.c can't be used
- Standard = snake_case : In production we user lowercase words seperated by underscores
- Contanst = UPPER_SNAKE_CASE : python doesn't enforce unchanable constants, so devs use this way to point out them
