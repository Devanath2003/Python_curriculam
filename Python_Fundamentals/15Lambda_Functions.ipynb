A lambda function in Python is a 
- small
- unnamed(anonymous) 
function written on a single line.

Whenever we need a quick, throwaway function - especially to pass into another function like sorted(), map(), or filter() - a lambda lets you define it right where we use it without cluttering our file with a full def block

## Syntax and Side by side comparsions with def 

Every lambda follow the exact same formula:

lambda parameters : single_expression


- that is there is no def keyword, no function name and no return keyword -Python automatically returns whatever the single expression evaluates to.

```python
# Standard function using def

def add_tax(price: float) -> float:
    return price * 1.73

# Exact same function written as a lambda

add_tax_lambda = lambda price : price * 1.73

print(add_tax(200.0))
print(add_tax_lambda(200.0))


# Note : Assigning a lambda to a variable like add_tax_lambda = ... is only done her to show
# how it works, In real code, PEP8 recommends using def if we are giving a function a name, and 
# reserving lambda for passing directly inside other functions.
```

## Verify of Basic Lambda Patterns 

- Multiple Parameters and Default values

just like def, a lambda can take zero, one or many parameters seperated by commas, including default values

```python
# Two arguments
calculate_offset = lambda page, limit: (page-1) * limit
print(calculate_offset(3,20))


# With a default parameter
format_url = lambda endpoint, version ="v1" : f"/api/{version}/{endpoint}"
print(format_url("users"))
print(format_url("users","v2"))
```

```python
# string manipulation

clean_slug = lambda title: title.strip().lower().replace(" ","-")
print(clean_slug("  FastAPI Backend Guide  "))
```

### conditional Logic inside a Lambda (Ternary Expression)

We cannot write a multi-line if/else block inside a lambda, but we can use a single line ternary expression (<valud> if <condition> else <value_if_else>)

```python
classify_status = lambda code: "Success" if 200<= code < 300 else "Error"

print(classify_status(201))
print(classify_status(500))
```

## Real World Use case: Customo sorting with 'key='

Python's sorted() function knows how to sort plain numbers [3,2,1] or strings ["b", "a"], but it doesnt' now how to sort a list of dictionaries or tuples unless we tell it which field to sort by

That is where lambda shines

```python
# Example A -Sorting a List of Tuples

# each tuple is (endpoint_path, avg_latency_ms)

endpoints = [
    ("/api/login/",180),
    ("/api/health",12),
    ("/api/reports",640),
    ("/api/users",95),
]

# By default , sorted() sorts by the First item (alphabetical path)
print("Alphabetical:", sorted(endpoints))

# Using lambda to sort by the SECOND item (index 1 -> latency)
fastest_first = sorted(endpoints, key = lambda item : item[1])
print("Fastest first: ", fastest_first)
```

```python
# Example B = Sorting a list of dictionaries (API JSON payloads)

users = [
    {"id": 103, "name": "charlie","role":"viewer", "login_count": 14},
    {"id":101,"name": "Alice", "roel":"admin", "login_count":82},
    {"id": 102, "name": "Bob", "role":"editor","login_count":45}
]

# Sort users by most active (descending login_count)

most_active = sorted(users, key = lambda user: user["login_count"], reverse =True)
for u in most_active:
    print(f"{u["name"]}: {u["login_count"]} logins")
```

```python
# Example C : Multi Level Sorting (Tie Breakers)

tasks = [
    {"title": "Fix DB index", "priority":1, "minutes": 45},
    {"title": "Update docs","priority":2, "minutes":15},
    {"title": "Patch auth bug","priority": 1, "minutes": 20}
]

# Sort by priority first (1 before 2) then by the shortest time (minutes)

ordered_tasks = sorted(tasks, key=lambda t: (t["priority"], t["minutes"]))
for task in ordered_tasks:
    print(task)
```

## Real world backend pattern: Dynamic Rule / Pricing Dispatch

In backend systems, lambdas are often stored inside dictionaries to create lightweight rule engines without writing long if/elif/else chains.


THIS IS VERY IMPORTANT AS WE CAN USE KEYS AS FUNCTIONNAMES AND CALL THEM EACH TO DO THEIR OWN SMALL FUNCTIONS WITH THE HELP OF LAMBDA

```python
# Discount calculators mapped by subscription tier

TIER_PRICING = {
    "free": lambda amount: amount,
    "pro": lambda amount: amount * 0.90,
    "enterprise": lambda amount : amount * 0.75,
}

user_tier = "pro"
cart_total = 2000.0

final_price = TIER_PRICING[user_tier](cart_total)
print(f"Tier {user_tier} carted out for : {cart_total} - final price : {final_price}")
```

## Rules and when not to use lambda

- single expression only: We cannot put loops (for,while) , variable assignments ( x = 5), or try/except blocks inside a lambda
- Readability comes first: If a lambda becomes hard to read on one line, convert it into a normal def function with a clear name and type hints

```python
routes = ["/api/v1/authentication/login", "/health", "/api/v1/users", "/docs"]

sorted_routes = sorted(routes, key = lambda endpoint : len(endpoint))
print(sorted_routes)
```
