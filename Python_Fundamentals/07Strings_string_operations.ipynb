A string (str) in Python is an immutable sequences of characters used to store and manipulate text.

Because strings are immutable, we cannot modify a character in place - every string operations returns a brand-new string

## Creation, Indexing and slicing

Stringes can use single quotes or double quotes or triple quotes for multi-line text.

- Because string is a sequqence , we can access individual characters by index or extract substrings using slicing. 

```python
endpoint = "/api/v1/users"

#Indexing (0 - based from start, negative from end)

print(endpoint[0])
print(endpoint[-1])

# Slicing [start:stop] (stop index is excluded)

print(endpoint[5:7]) 
print(endpoint[8:]) # from index 8 to the end
print(endpoint[:4]) # from start up to index 4

# Immutability check
endpoint[0] = "x" # It raises TypeError 'str' object does not support item assignment
```

## Modern String Fromatting (f-strings)

Formatter string literals (f"") are the standard way in python to embed varaibles and expressions directly inside text without manually calling str()

```python
service = "auth-service"
latency_ms = 14.3876
status = 200

# Injecting variables, expressions, and float precision formatting
log_entry = f"[{service.upper()}] Status: {status} | Latency: {latency_ms:.2f}ms"
print(log_entry)

```

## Essential String Methods

String methods never alter the original string; always assign the result back to a varaible if you want to keep the change

```python
raw_input = "   Admin@Company.COM  \n"

# Cleaning and normalizing

cleaned = raw_input.strip().lower()
print(cleaned)

print(cleaned.endswith("@company.com")) # check whether it endswith the specific string
print(cleaned.startswith("admin"))


# Replace substrings
masked = cleaned.replace("admin", "*****")
print(masked)


#Splitting a string into list and joining a list back into a string 
csv_roles = "viewer,editor,admin"
role_list = csv_roles.split(",")
print(role_list)

formatted_roles = " | ".join(role_list)


print(formatted_roles)
```

# Real world Backend and API Examples

```python
# Example A : Extracting a JWT from an HTTP Authorization Header

auth_header = "Bearer eyJhbsadfgHkdsfgiohsdglkhglLKHK7hKLH08KKH"

if auth_header.startswith("Bearer "):
    # Split on space once and take the second part (the token)
    token = auth_header.split(" ", maxsplit=1)[1]
    print("Extracted Token:", token)
else:
    print("401 Unauthorized: Invalid header format")
```

## Safe prefix and Suffix Removal (removeprefix / removesuffix)

A common bug in Python using .strip() to remove a word, because .strip("api") removes any combination of character 'a', 'p' , and 'i' from the ends.

so removeprefix and removesuffix remove exact substrings safely.

```python
route = "/api/v1/items.jsoni"

#Wrong
wrong_out = route.strip("/api") # it removing the api from the front also the i at the end

# Right

clean_route = route.removeprefix("/api/v1") # it only removes from the front
endpoint_name = clean_route.removesuffix(".jsoni") # it only removes from the end

print(wrong_out)
print(clean_route)
print(endpoint_name)
```

## Searching and Validation Methods

Before converting strings or querying a database, we often need to inspect what characters a string contaiins or locate specific markers.

```python
log_line = "ERROR: Connection timeout on port 5432 (timeout)"

#Searching

print(log_line.count("timeout")) # count how many times does this word appears
print(log_line.find("5432")) # returns starting index or -1 if not found
print(log_line.find("9999")) # Safe output as -1, index() will raise an error


# character validation (always returns bool)

otp_code = "3231"
username = "devanth030"

print(otp_code.isdigit()) # checks if its all digits
print(username.isalnum()) # checks if it is alpha numerical
print("   \t\n".isspace()) # Checks only whitespace
```

## Escape characters and Raw strings (r"...")

Backslashes (\) trigger special characters like newlines (\n)or tabs (\t).
- Prefixing a string with 'r' creates a raw string that treats backslashes literally - essential for regular expressions, file paths, and SQL/log patterns. 

```python
# Standard string interprets \n and \t
formatted_text = "Line 1\n\tIndented Line 2"
print(formatted_text)

# we making the same string ignore the special characters
formatted_text = r"Line 1\n\tIndented Line 2"
print(formatted_text)

# Raw string ingores escape sequences

regex_pattern = r"^\d{3}-\w+\n$"
print(regex_pattern)
```

# Strings (str) vs Bytes (bytes) = Encoding and Decoding

Netoworks, HTTP requests, cryptographic hashing, and JWT tokens do not transmit Python str objects  - they transmit raw bytes (prefixed wiht b"...")

- Converting bw text and binary data uses.encode() and .decode()

```python
secret_payload = "user_id.42|role:admin"

# str ->bytes (UTF-8 is the universal web standard)
raw_bytes = secret_payload.encode("utf-8")
print(raw_bytes)
print(type(raw_bytes))

# bytes -> str

decoded_text = raw_bytes.decode("utf-8")
print(decoded_text)

```

## Padding and Formatting (zfill and f-string alignment)

Useful for generating sequential invoice numbers, database IDs, or clean terminal logs

```python
invoice_num = "47"
print(invoice_num.zfill(6)) # Pads with leading zeros to reach length of 6

# Aligning columns in terminal output with f-strings
method = "POST"
path = "/api/v1/login"
print(f"{method:<5}| {path:<20} | 200 OK")
```

## Real world backend & API examples

Combining these tools lets parse structured raw strings - such as query string or webhook headers - cleanly wihtout external libraries

```python
# Example A - Parsing a Raw URL Query string

raw_url = "https://api.example.com/v1/products?category=laptop&sort=price_asc"

if "?" in raw_url:

    # Split into base URL and query string at the first '?' 
    base_url, query_string = raw_url.split("?", maxsplit=1)
    clean_path = base_url.removeprefix("https://api.example.com")
    print(f"Endpoint: {clean_path} | Raw Query: {query_string}")

    
```

```python
# Example B - Validating an Incoming API Key Format

incoming_key = "sk_live_982374928374"

is_valid_key = (incoming_key.startswith("sk_live")
                and incoming_key.removeprefix("sk_live_").isdigit()
                and len(incoming_key)== 20)

print("API key valid:",is_valid_key)
```
