Loops allow program to execute a block of code repeatedly , either once for each item in a sequence (for loop) or as long as a condition remains True (while loop)

## The for loop & range()

A for loop iterates over a sequence of items. 

When you need to repeat an action a specific number of times, use Python's built-in range() function (note that the stop value is always excluded)

```python
# range(stop) 0 -> stop-1

for i in range(5):
    print("Attempt:", i)

# range(start, stop, step) -> even numbers from 2 to 8
for num in range(2,10,2):
    print(num)

# Iterating directly over characters in a staring

service_name = "API"

for char in service_name:
    print(char)

# using enumerate() when you need BOTH the index and the value
for index, chart in enumerate(service_name, start=1):
    print(f"Character {index}: {chart}")
```

## The while loop

A while loop keeps running as it condition evaluates to True.

Must update a variable inside the loop so the condition eventually becomes False; otherwise , it leads to an infinite loop

```python
queue_size = 3
while queue_size > 0:
    print(f"Processing job... {queue_size} remaining")
    queue_size -= 1

print("All jobs processed")
```

## Loop Control (break, continue, and else)

- continue : Skips the rest of the current iteration and jumps staright to the next cycle
- break : Immediately exits the loop completely
- else : on a loop : An optional else block after a for or while loop runs only if the loop finished naturally without hitting a break


```python
# Skipping invalid items with 'continue' and stopping early with 'break'

for code in [200,201,400,500,200] :
    if code == 400:
        print("Skipping bad request(400)")
        continue
    if code == 500:
        print("Critical server error(500)! Aborting batch.")
        break
    print(f"Logges successful status: {code}")
```

## Real world Backend & API Examples

In backend services, loops power network retry mechanisms and paginated data fetching.

```python
# Example A : API retyr logic iwth 'for...else' ---

MAX_RETRIES = 3

for attempt in range(1, MAX_RETRIES +1):
    print(f"Calling payment gateway (Attempt {attempt}/{MAX_RETRIES})...")

    # Simulating success on the 2ns attempt
    simulated_status = 200 if attempt == 2 else 503

    if simulated_status == 200:
        print("Payment confirmed!")
        break
    print("Service unavailable (503), Retyring...")

else:
    # Runs only if 'break' was never triggered across all attempts
    print("Error: Payment gateway unreachable after max retries.")
```

```python
# Example B : Paginated database batch fetching

total_records = 55
batch_size = 20
offset = 0

while offset < total_records :
    remaining = total_records - offset
    fetched = batch_size if remaining >= batch_size else remaining
    print(f"Fetched {fetched} rows starting at offset {offset}")
    offset += fetched
```
