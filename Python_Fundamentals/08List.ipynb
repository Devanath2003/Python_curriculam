A list in Python is 
- Ordered
- Mutable collection

used to store multiple items inside a single variable. Unlike strings, lists are mutable - meaning we can add, update, reorder, or remove items inplace without creating a new list

## Creating, Indexing and Slicing

- Created using square brackets
- Can hold elemements of any type (including mixed types or other lists)
- Indexing and slicing work identically to strings

```python
# Creating lists (with optional type annotations)
http_methods : list[str] = ["GET", "POST", "PUT", "DELETE"]
scores: list[int] = [88, 92, 79, 95, 84]

#Accessing and updating items by index
print(http_methods[0])
print(http_methods[-1])

# Mutating (changing an item in place)
http_methods[2] = "PATCH"
print(http_methods)

# Slicing [start:stop] returns a new sublist
top_three = scores[:3]
print(top_three)
```

# Adding and removing elements
builtin methods to modify lists dynamically

```python
tasks = ["db_backup", "send_emails"]

# Adding items
tasks.append("cleaup_logs") # Adds to the end
print(tasks)
tasks.insert(0, "health_check") # Inserts at index 0
print(tasks)

tasks.extend(["sync_cache", "audit"]) # Appends multiple items from another list
print(tasks)
```

```python
# removing items

last_task = tasks.pop() #Remove and returns the last item
print(last_task)
first_task = tasks.pop(0) #Remove and returns the item at index 0
print(first_task)
tasks.remove("send_emails") # Remove the first matching value (raise ValueError else)

print(tasks)
```

## Sorting, Reversing and Useful Helpers

There is a close difference bw 
- .sort() - Modifies the list directly and returns None
- sorted() - Leaves the original list untouched and returns new sorted list.

```python
latencies_ms =[142,38,210,65]

print(len(latencies_ms))
print(min(latencies_ms), max(latencies_ms)) 
print(38 in latencies_ms) # Check if it is present

# sorted() create a new list; original remains unchanged

ordered_latencies = sorted(latencies_ms)
print("Sorted Copy",ordered_latencies)
print("Original", latencies_ms)

# .sort() mutates the list in place


latencies_ms.sort(reverse=True)
print("In-Place descending",latencies_ms)
```

## Real world Backend and API examples

in FastAPI and database workflows, lists represent collections of returned records, middleware stakcs, and paginated response slices

```python
# Example A - Slicing Database results for pagianation

all_user_ids = [101,102,103,104,105,106,107,108,109,110]
page = 2
page_size = 3

start_idx = (page -1) * page_size
end_idx = start_idx + page_size


page_data = all_user_ids[start_idx: end_idx]
print(f"Page {page} records:",page_data)

```

```python
# Example B  - Tracking Valid vs Failed Records in a Batch job

incoming_emails = ["alice@test.com", "invalid-email","bob@test.com", "bad_format"]
valid_emails: list[str] = []
failed_emails: list[str] = []

for email in incoming_emails:
    if "@" in email and "." in email:
        valid_emails.append(email)
    else:
        failed_emails.append(email)

print("Queued for invite:", valid_emails)
print("Rejected payloads:", failed_emails)
```
