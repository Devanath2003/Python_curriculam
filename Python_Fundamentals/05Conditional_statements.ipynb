## if , elif, and else structure

python uses colon : and tab or 4 spaces as indentation.

it evaluates conditions from top to bottom and executes only the first branch that evaluates to True.

```python
response_time = 320

if response_time < 100:
    print("Performance: Faset")
elif response_time < 500:
    print("Performance: Acceptable")
elif response_time < 1000:
    print("Performance: Slow")
else:
    print("Performance: Critical Timeout")
```

## Python Truthiness and Ternary Expressions

- Direct Truthiness checks : Because empty strings (""), 0, and None evaluate to False, Python developers write 

    if token:

    rather than , if token != "" and token is not None:

- Ternary Operator (inline if/else) : Ideal for assigning a variable based on a simple binary condition on a single line

```python
# Checking truthiness directly

api_key = ""

if not api_key:
    print("API key is missing")

# Ternary Expression  : <value if_true> if <conditiona> else <value_if_false>

is_production = False
log_level = "Error" if is_production else "Debug"
print("Log level:", log_level)
```

## Real world bakckend and API examples

in FastAPI service, conditonal logic hadles HTTPS status routing, payload validation, and user access rules.

```python
# Example - A : Handling upstream PAI Status codes ----

status_code = 429

if 200 <= status_code < 300:
    print("Request succeeded")
elif status_code == 401 or status_code == 403:
    print("Auth error: Invalid credentials or insufficient permissions")
elif status_code == 404:
    print("Resource not found.")
elif status_code == 429:
    print("Rate limits exceeded . Retry after backoff")
else:
    print("Server error:", status_code)
```

```python
# Example B - Request validation and Access Gatekeeping

user_role = "intern"
is_account_active = True
payload_size_kb = 120
MAX_PAYLOAD_KB = 500

if not is_account_active :
    print("403 Forebidden: Account suspended")
elif payload_size_kb > MAX_PAYLOAD_KB:
    print("412 payload too large")
elif user_role == "admin" or user_role == "intern":
    print("200 OK: Access granded to sandbox endpoint")
else:
    print("403 Forebidden: Role not permitted")
```
