Comprehensions are a concise , expressive python syntax for building a new list, dictionary, or set in a single readabale line by transoforming and/or filtering items from an existing iterable.

In professional Python and FastAPI codebases , comprehensions are preferred over map() and filter() because they do both jobs at once without needing lambda or list() wrappers.

## List comprehensions 
formula:
[expression for item in iterable if conditon]

Read it left to right like English : Give me [expression] for each item in iterable (optionally if condition is true)

```python
# Basic transformatio (Replaces map())

# Traditional for-loop way:
endpoints = []
for r in ["users","orders","items"]:
    endpoints.append(f"/api/v1/{r}")

# List comprehension way:
resources = ["users","orders", "items"]
endpoints = [f"/api/v1/{{r}}" for r in resources]
print(endpoints)
```

```python
# B. Filtering only replaces filter()

status_codes = [200,404,201,500,403, 204]

error_codes = [code for code in status_codes if code>=400]
print(error_codes)
```

```python
# C. Transforming AND filtering Together

raw_emails = ["  AiceTest.com ", "BOB@COMPANY.COM", None, "  charlie@dev.io "]

# Strip and lowercase ONLY non_empty strings

clean_emails = [email.strip().lower() for email in raw_emails if email]
print(clean_emails)
```

# D. Conditional Expression (if/else) inside the Transofrmation


Notice where if goes depending on what we want:

- if at the END : filters out items (reduces list  lenght)
- if/else at the START: Transoforms every item differently (keeps the exact same list length)

```python
latencies = [45, 320, 88, 610]

labels = ["SLOW" if ms > 200 else "FAST" for ms in latencies]
print(labels)
```

# Flattening Nested Lists (Two for clauses)

```python
role_permissions = [
    ["read:profile", "update:profile"],
    ["read:orders", "create:orders"],
    ["delete:users"],
]

# Outer loop comes first, inner loop comes second

all_permissions = [perm for group in role_permissions for perm in group]
print(all_permissions)
```

## Dictionary Comprehensions

formula:

{key_expr : value_expr for item in iterable if condition}

```python
# A. Building and ID- indexed lookup map (Essential Backend Pattern)

users = [
    {"id": 101, "username": "devanath", "active": True},
    {"id": 102, "username": "alice", "active": False},
    {"id": 103, "username": "bob", "active": True},
]

# Map user_id -> username for acitve users only

active_user_map = {u["id"] : u["username"] for u in users if u["active"]}
print(active_user_map)
```

```python
# B. Transofroming or filtering a dictionary

prices_usd = {"basic": 10, "pro": 25, "enterprise": 100}

# Converts all plans to INR (assuming 1 USD = 84 INR)

prices_inr = {plan: amount * 84 for plan, amount in prices_usd.items()}
print(prices_inr)

# keep only paid plans above 15 doller

premium_plans = {plan : amt for plan, amt in prices_usd.items() if amt > 15 }
print(premium_plans)
```

```python
# C. Inverting keys and values

status_messages = {200: "OK",404: "NOT FOUND", 500: "SERVER ERROR"}

code_by_message = {msg: code for code, msg in status_messages.items()}
print(code_by_message)
```

## Set comprehensions {..}

{expression for item in iterable if condition}

```python
log_entries = [
    "2026-11-18 | auth-service | INFO",
    "2026-11-18 | billing-service | ERROR",
    "2026-11-18 | auth-service | WARNING",
    "2026-11-18 | user-service | INFO",
]

# Extract unique service names from raw log strings

unique_service = {line.split("|")[1].strip() for line in log_entries}
print(unique_service)
```

## Real world FastAPI and Bakckend example

when an API endpoint returns a listof database recrods, comprehensions cleanly serialize and sanitize the response payload before returning JSON.



```python
db_orders = [
    {"order_id": 501, "customer": "Devanath", "total": 1499.0, "paid": True},
    {"order_id": 502, "customer": "Alex", "total": 299.0, "paid": False},
    {"order_id": 503, "customer": "Priya", "total": 3200.0, "paid": True},
]

# Serialize only paid orders into lightweight API summary objects

paid_order_summaries = [{"id" : o["order_id"], "amount_with_tax": round(o["total"] *1.18,2)} for o in db_orders if o["paid"]]

print(paid_order_summaries)
```

# GOlden Rule of Readability

Keep comprehensions to one transformation and one simple condtion. If we find ourself writing nested if/else chains or multiple nested for loops that take effort to read,write a standard for loop instead.

```python
products = [
    {"name": "Mouse", "price": 650},
    {"name": "Keyboard", "price": 2400},
    {"name": "USB Cable", "price": 250},
]

cheap_products = [pro["name"] for pro in products if pro["price"] < 1000 ]

print(cheap_products)
```
