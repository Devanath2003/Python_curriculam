A tuple in python is an 
- ordered
- immuatable collection
of items. Think of it as a locked list: once a tuple is created , we cannot add, remove or replace its elements

## Creation and The single element comma rule
Tuples are typically written in parenthesis (..).

though it is acutally the comma that creates a tuple. 

A commong begineer bug is forgetting the trailing comma when creating a tuple with just one item

```python
# Standard tuples
http_ports : tuple[int,int] = (50, 443)
db_endpoint : tuple[str, int] = ("127.0.0.1", 5432)

#Indexing and slicing work just like lists and strings
print(db_endpoint[0])
print(http_ports[-1])

```

```python
# THE SINGLE ELEMENT GOTCHA

not_a_tuple = (200) # Python treats this as a plain int: 200
valid_tuple = (200,) # The trailing comma makes it a 1-element tuple

print(type(not_a_tuple))
print(type(valid_tuple))
```

## Why use Tuples instead of Lists ?

- Data Integrity: Protects fixed records(like database rows,coordinates, or server configurations ) from accidental modification elsewhere in the code.
- Performance: Tupules use less memory and are faster to create than lists because Python does not need to allocated extra space for future .append() calls
- Usable as Dictionary Keys: Because tupules are immutable , they can be uses as keys in a dictionary (lists cannot)

```python
# Attempting to modify a tuple raises a TypeError
allowed_roles = ("admin","editor","viewer")
# allowed_roles[0] = "superadmin" #Error

#Only two built in methods exist since we can't mutate it:
print(allowed_roles.count("admin"))
print(allowed_roles.index("viewer"))
```

## Tuple unpacking (A python superpower - ??)

the values of a tuple can be unpacked directly into seperate varaibles in a single line.

we does this constantly when 
- iterating over database rows
- calling functions that return multiple values

```python
# Unpacking exact elements 

server_config = ("0.0.0.0", 8000,"production")
host, port, env = server_config
print(f"Running on {host}: {port} [{env}]")
```

```python
#Extended unpacking with * (captures remaining items into a list)

status_codes = (200,201,400,404,500)
first_ok, second_ok, *error_codes = status_codes
print(first_ok)
print(error_codes)
```

## Real world Backend & API Examples

Relations database and SQL drivers return each fetched row as a tuple because a single row has a fixed column structure.

```python
# Example A : Iterating over Raw Database Rows

#Simulated rows returned from a PostgreSQL query: (user_id,username, is_active)

db_rows: list[tuple[int,str,bool]] = [(1,"devanath",True),(2,"alex",False),(3,"priya_k",True),]

#Unpacking each row tuple directly inside the loop header

for user_id,username,is_active in db_rows:
    if is_active:
        print(f"ID {user_id}: {username} is acitve")

```
