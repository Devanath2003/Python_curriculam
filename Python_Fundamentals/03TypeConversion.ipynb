Type conversion  also called Type casting - is the process of changing value from one data type into anther, either automatically by python on manually using built-in constructor functions.

## Implicit Type Conversion (Automatic)

when we try to combine compatible types in an expression, Python automatically converts the "smaller" types into the "larger" type to prevent data loss.

```python
# int + float -> python automatically promotes the result to float

item_count = 3 
shipping_fee = 234.4

total = item_count + shipping_fee

print(total)
print(type(total))

#int + bool -> True is treated as 1 and False is treated as 0

base_score = 12
bonus = True
print(base_score + bonus)
```

## Explicit type conversion (Manual Casting)

When Python Cannot convert types automatically (like adding an str and an int) , we can convert them explicitly using int(), float(), str(), or bool()

```python
# conversion of strings into numbers
raw_age = "26"
age = int(raw_age)

raw_price = "134.5"
price = float(raw_price)

# converting numbers to string s(required for concatenating)
status_code = 200
message = "HTTP stats: "+ str(status_code)
print(message) 


# float -> int truncates or chopss off the decimals , IT DOES NOT ROUND
exact_rating = 3.43
truncated_rating = int(exact_rating)
print(truncated_rating)

```

## Common conversion pitfalls

- Decimal strings to int :int("123.5") crashes with valueError. for that , i need to convert it into float first then into int as (int(float(2.45))) 
- converting any empty or non zero values (eg -  0, 0.0, "", None) returns False, but any non empty string evaluates to True even the word "False" 

```python
print(bool(0))
print(bool(""))
print(bool(None))

# all aboove false

print(bool("Hello"))
print(bool("0"))
print(bool("False"))
```

## Real - world backedn & API examples

In backend engineering , data entering the application from Linux enviro variables, URL query strings, or terminal inputs almost always arrives as str -> and it must be cast berore use

```python
env_port = "5050"
env_timeout = "14.5"
env_debug = "false"

# Converting to usable backend types

server_port: int = int(env_port)
request_timeout: float = float(env_timeout)

## Safe string to boolean conversion - never use bool(env_debug) directly.

is_debug : bool = env_debug.lower() == "true"
print(server_port, request_timeout, is_debug)



# Example B - Processing user terminal output

simulated_input = "4"
unit_price = "235.2"
order_total = int(simulated_input) * float(unit_price)
print("Order Total:" + str(order_total))
```
