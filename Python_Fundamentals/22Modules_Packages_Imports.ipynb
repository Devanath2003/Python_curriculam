Modules,Packages , and imports allow you to split a large application into clean, reusable files instead of stuffing thousands of lines of code into a single script.

In FastAPI projects, this is how routes, database models, authentication helpers, and configurations settings are stay organized.

## Modules and import styles

A module is simply any .py file containing python functions, variables, or classe. 

Every .py file we create is also a module we can import into other files

### Four ways to import:

```python
# 1. Import the entire module (clear origin , zero name collisions)
import math
print(math.ceil(4.3))

# 2. Import with an alias using 'as' (common for long module names)
from datetime import datetime as dt
print(dt.now().year)

# 3 Import specific items directly (cleanes for frequently used functions)
from pathlib import Path
config_path = Path("pyproject.toml")

# 4. Wildcard import : From math import *
# AVOID THIS: Pollutes our file's namespace and hides where functions came from.
```

## Packages and __init__.py

A package is a folder containing multiple related .py modules grouped together. Remember the sr/python_fundamentals/__init__.py file that uv init generated in our folder earlier

- Placeing an __init__.py file (even if completely empty) inside a directory explicitly marks that folder as python package.

- We can also use __init__.py to expose key functions at the package level so other files have shorter,cleaner import paths:

### Typical Modular Backend Folder sturcture

my_backend/

├── main.py                  # Entry point of the application

└── app_utils/               # Package directory

     ├── __init__.py          # Marks 'app_utils' as a package

     ├── validators.py        # Module for email/token validation

     └── formatters.py        # Module for response formatting

## The if __ name __ == "__ main __" : Guard

whenever Python imports a module, it executes every top-level line of code inside that file from top to bottom

To prevent test code or print() statements inside a helper module from running automatically when another file imports it, wrap script execution code inside if __ name __ == "__ main __"


- when we run uv run validators.py directly , Python sets __ name __ to "__ main __" , so the block runs

- when main.py runs from app_utils.validators import is_valid_email, Python set __ name __ to "app_utils.validators" , so the test block is skipped.

```python
# File: app_utils/validators.py

def is_valid_email(email:str) -> bool:
    return "@" in email and "." in email.split("@")[-1]


# Only runs if we execute this file directly - NOT when imported

if __name__ == "__main__":
    print("Running standalone validator tests..")
    print(is_valid_email("devanath@company.com"))
```
