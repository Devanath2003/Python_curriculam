zip () is a builtin python function  that takes two or more iterables (like lists or tuples) and paris their elements together position- by position into tuples, just like the teeth of zipper closing together

## Syntax and how zip() works

zip(iterable1, iterable2)

just like map(), filter(), zip() returns a lazy zip iterator. Wrapping it in list(..) or looping over it directly reveals the paired tuples

```python
user_ids = [101,102,103,]
usernames = ["devanath", "alice", "bob","raj"]

# Pairs index 0 with index 0, index 1 with index 1, and so on 
paired = list(zip(user_ids, usernames))
print(paired)


# we can zip three or more lists at the same time

roles = ["intern", "admin", "viewer"]
full_records = list(zip(user_ids, usernames, roles))
print(full_records
)
```

## Variety of Essential zip() patterns 

Example A - Looping over parallel lists cleanly (No Index Math)

instead of writing for i in range(len(endpoints)):,

zip() lets us unpack both lists directly in the loop header

```python
endpoints = ["/api/login", "/api/users", "/api/orders"]
status_codes = [200, 200, 503]

for route, status in zip(endpoints, status_codes):
    print(f"{route:<15} -> HTTP {status}")
```

## Example B: creatint a dictionary instantly with dict(zip())

Passing a zip() object directly into dict() is one of the fastest, cleanest ways in Python to turn a list of keys andd list of values into a dictionary

```python
headers = ["id", "username","plan"]
row_values = [42, "devanath_dr","pro"]

user_dict = dict(zip(headers,row_values))
print(user_dict)
```

## Example C : "Unzipping" paired data back into seperate lists (zip(*..))

Using the * unpacking operator inside zip() reverses the process , splitting a list of tuples back into individual tuples

```python
pairs = [("GET", 120), ("POST", 45), ("DELETE", 8)]

methods, counts = zip(*pairs)
print("Methods", methods)
print("Counts",counts)
```

## Handling unequal list lengths (strict = True)

by default , zip() stops as soon as the shortest iterable runs out of items, silently ignoring any leftover elements in longer lists

but in 3.10 python strict=True tells Python to raise a ValueError if the lengths do not match - preventing silent data loss

```python
keys = ["host","port","debug"]
values = ["127.0.0.1",8000] #missing a third value here

# Default behaviour - silently drops "debug"
print(dict(zip(keys,values)))


# safe behaviour in backend pipelines; raise Value Error when lengths mismatch

print(dict(zip(keys,values,strict=True)))
```

## Real - world Backend and API Examples

Example A : converting raw SQL query results into JSON Dictionaries

Low-level database drivers often return column names seperately from row tuples. Combining zip() with map() converts raw rows into FastAPI- ready JSON dictionaries in two lines

```python
columns = ["product_id", "name", "in_stock"]

db_rows = [
    (1,"Mechanical Keyboard", True),
    (2, "USB-C-Dock", False),
    (3,"4K-Monitor", True),
]

json_products = list(map(lambda row: dict(zip(columns, row)),db_rows))
print(json_products)
```

```python
# Example B : Detecting changes between two states (Audit log diffing)

fields = ["username","role","email"]
old_state = ["devanath","intern", "dev@old.com"]
new_state = ["devanath","bakcend_dev", "dev@company.com"]

for field,before,after in zip(fields,old_state, new_state, strict=True):
    if before != after:
        print(f"Field '{field}' changed: '{before}' -> '{after}'")
```
