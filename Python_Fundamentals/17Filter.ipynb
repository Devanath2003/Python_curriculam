filter() is a built in python function that tests every item in an iterable against a condition and keeps only the items for which the condiiton returns True.

Think of filter() as a security bouncer at a door, every item shows its ID (passes through our test function), items that pass(True) get let through, and items that fail(False) are dropped. Unlike map(), which changes what items look like without changingthe count, filter() leaves items untouched and reduces how many survive

## Syntax and How filter() works

filter(condition_function, iterable)

just like map(), filter() returns a lazy filter iterator , so wrap it in list() to collect and inspect the surviving items

```python
status_codes = [200,404,201,500,204,403]

# Keep only successful 2xx HTTP status codes
successful_codes = list(filter(lambda code: 200<=code<=300,status_codes))
print(successful_codes)
```

```python
# filtering strings by pattern

endpoints = [
    "/api/v1/users",
    "/api/v2/users",
    "/api/v1/orders",
    "/health",
    "/api/v2/payments",
]

v2_routes = list(filter(lambda path: path.startswith("/api/v2"),endpoints))
print(v2_routes)
```

Example - THe filter(None, iterable) Shortcut

If we pass None as the first argument instead of a function, Python automatically filters out all falsy values ("",None,0,False,[]) and keeps only truthy ones

```python
raw_tags = ["python","",None,"fastapi","  ","postgres",0]

# Removes Falsy values but note that "  " is truthy
non_empty = list(filter(None,raw_tags))
print(non_empty)
```

## Real world Backend and API examples





```python
# Example A : Filtering Active Admin records from a database result

users = [
    {"id": 1, "name": "Alice", "role": "admin", "is_active": True},
    {"id": 2, "name": "Bob", "role": "viewer", "is_active": True},
    {"id": 3, "name": "Charlie", "role": "admin", "is_active": False},
    {"id": 4, "name": "Devanath", "role": "admin", "is_active": True},
]

active_admins = list(filter(lambda user: user["role"]=="admin" and user["is_active"],users))
print(active_admins)
```

## Example B : Chaining filter() and map() Together(Data pipeline)

In real backend code, we frequently filter records first to select what matters, and then map over the survivors to extract or format the output

```python
# Goal : Get a list of uppercase emails for ONLY active admin users
active_admin_emails = list(
    map(lambda u: u["name"].lower(),filter(lambda u:u["role"] == "admin" and u["is_active"], users))

)

print(active_admin_emails)
```

```python
server_logs = [
    "INFO: Server started on port 8000",
    "ERROR: Database connection refused",
    "INFO: GET /health 200 OK",
    "WARNING: High memory usage (82%)",
    "ERROR: JWT signature verification failed",
]

errors = list(filter(lambda log: log.startswith("ERROR"),server_logs))
print(errors)
```
