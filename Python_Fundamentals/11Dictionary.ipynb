A dictionary (dict) in Python is a
- mutable
- insertion-ordered collection of key-value pairs

Instead of looking up itembs by a numberic position like a list, we can look up values using a unique, immutable key (such as str, int or tuple) in near instataneous O(1) time

making dictionaries the direct Python equivalent of JSON objects

## Creation, Accessing , and safe Lookups (.get())

Idctionaries are written with curly braces {key:value}. Using square brackets user["key"] crashes with a KeyError if the key does not exist

whereas .get("key",default) safely returns None (or your chose fallback)

```python
# Creating a dictionary

user_profile: dict[str, str | int | bool] = {
    "id" : 101,
    "username" : "devanath-dr",
    "role" : "intern",
    "is_active": True,
}


# direct lookup (fast, but raise KeyError if missing)

print(user_profile["username"])

# Safe lookup with .get() (prefered when handling external API payloads)

print(user_profile.get("email"))
print(user_profile.get("email","not_provided")) # give fallback value

# checking if a key exists (always checks keys, NOT values)
print("role" in user_profile)
print("inern" in user_profile)
```

```python
config = {"host" : "127.0.0.1","port":8000}

# Adding a new key and updating an existing key

config["debug"] = True
config["port"] = 8080


# Merging two dictionaries (Python 3.9+), '|' operator; right side overrides left

override_config = {"port":9000, "workers": 4}
final_config = config | override_config 

print(final_config)


#Removing keys
removed_workers = final_config.pop("workers")
safe_remove = final_config.pop("non_existent",None) # safe pop with default - avoids keyerror
print(final_config)
```

## Iterating over dictionaries (.items())

we can loop through keys, values or both simultaneously using .items() combined with tuple unpacking

```python
stats_counts = {200:1450, 404:32, 500:3}

# Iterating over keys and values together

for status_code, count in stats_counts.items():
    print(f"HTTP {status_code}: {count} responses recieved")
```

## Real wordl backend and API examples

in FastAPI, every incoming JSON body and outgoing API response is structured as nested dictionaries and lists, 

and dictionaries are frequently used to index database records by ID for fast lookups

```python
# Example A - Nested JSON API Response structure

api_response = {
    "status" : "success",
    "data" : {
        "user" : {"id":42, "name":"Devanath"},
        "permissions" : ["read:repo", "write:code"],
    },
    "meta" : {"request_id": "req_89a7f"},
}

first_perm = api_response["data"]["permissions"][0]

print("Primary permission",first_perm)

```

```python
# Example B = Indexing records by ID to avoid nested loops
# Converting a list of user dicts into an ID-indexed lookup map

users_list = [
    {"id":10, "name":"Alice",},
    {"id":20, "name":"Bob"},
    {"id":30, "name":"Charlie"},
] 

users_by_id: dict[int, dict[str,int | str]] = {}

for user in users_list:
    users_by_id[user["id"]] = user

# Instant O(1) lookup by ID without scanning the list
print("Found user #20:", users_by_id.get(20))
```
