## Encapsulation and Naming conventions (_protected vs __private)

Python does not enforce strict private keywords like Java or C++, instead, it uses underscore prefixes to signal intent:

- Public (self.name) = Safe to read and modify directly from outside the class

- Protected(self._timeout) = single leading underscore signlas "internal implementation detail - do not modify directly outside the class or its subclasses."

- Private(self.__secret_key) = Double leading underscore triggers name mangling (Python renames it internally to _ClassName__secret_key), preventing accidental overrides or casual external access.


```python
class APIClient:
    def __init__(self, base_url: str, api_key: str) -> None:
        self.base_url = base_url # Public
        self._retry_count = 3 # Protected (internal convention)
        self.__api_key = api_key # Private (name - mangles)
    
    def get_masked_key(self)  -> str:
        return f"{self.__api_key[:4]}****"

client = APIClient("https://api.example.com", "sk_live_987654")
print(client.get_masked_key())
```

## Proterties (@proerty and @<name>.setter)

In Python , we don't write ugly get_port() and set_port() methods.

we start with normal attributes(server.port = 8080), and if we later need valiation or computed visuals, we wrap them in @property, without changing how callers use our class

```python
class ServerConfig:

    def __init__(self, host:str, port:int) -> None:
        self.host = host
        self.port = port # Triggers the @port.setter below!

    # Getter: runs ehen someone reads config.port

    @property
    def port(self) -> int:
        return self._port
    
    #runs 
```
